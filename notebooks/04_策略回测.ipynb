# 策略回测

回测双均线策略和 RSI 策略的历史表现。

```python
import sys
sys.path.insert(0, '..')

import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from src.analysis.strategies import MovingAverageCrossStrategy, RsiStrategy
```

```python
# 获取数据
df = yf.Ticker("AAPL").history(period="1y")
df.columns = [c.lower() for c in df.columns]
close = df['close']
```

```python
# 双均线策略回测
ma_strategy = MovingAverageCrossStrategy(fast=20, slow=50)
signals = ma_strategy.generate_signal(df)

df['signal'] = signals
buy_signals = df[df['signal'] == 1]
sell_signals = df[df['signal'] == -1]

print(f"买入信号: {len(buy_signals)} 次")
print(f"卖出信号: {len(sell_signals)} 次")
```

```python
# 绘制
fig = go.Figure()
fig.add_trace(go.Scatter(x=df.index, y=close, name='Close', line=dict(color='black')))
fig.add_trace(go.Scatter(x=df.index, y=close.rolling(20).mean(), name='MA20', line=dict(color='orange')))
fig.add_trace(go.Scatter(x=df.index, y=close.rolling(50).mean(), name='MA50', line=dict(color='blue')))
fig.add_trace(go.Scatter(x=buy_signals.index, y=buy_signals['close'],
    mode='markers', name='Buy', marker=dict(color='green', size=10, symbol='triangle-up')))
fig.add_trace(go.Scatter(x=sell_signals.index, y=sell_signals['close'],
    mode='markers', name='Sell', marker=dict(color='red', size=10, symbol='triangle-down')))
fig.update_layout(title='AAPL 双均线策略信号', height=500)
fig.show()
```
