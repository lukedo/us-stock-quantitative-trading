# 期权定价与 Greeks

使用 Black-Scholes 模型计算期权理论价格和 Greeks。

```python
import sys
sys.path.insert(0, '..')

from src.analysis.options_pricing import (
    black_scholes_price, calculate_greeks, implied_volatility
)
import pandas as pd
```

```python
# Black-Scholes 定价示例
S = 200      # 标的当前价
K = 205      # 行权价
T = 30/365   # 到期时间(年)
r = 0.045    # 无风险利率
sigma = 0.25 # 波动率

call_price = black_scholes_price(S, K, T, r, sigma, 'call')
put_price = black_scholes_price(S, K, T, r, sigma, 'put')
greeks = calculate_greeks(S, K, T, r, sigma, 'call')

print(f"Call Price: {call_price:.2f}")
print(f"Put Price: {put_price:.2f}")
print(f"Greeks: {greeks}")
```

```python
# 隐含波动率计算
market_price = 6.50
iv = implied_volatility(market_price, S, K, T, r, 'call')
print(f"Implied Volatility: {iv:.2%}")
```
